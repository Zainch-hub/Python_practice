# creating a list of products


```python
products = [ 'Mobile' , ' Laptop ' , ' headphones ' , ' books ' , ' pen ' , ' charger ' ] 
```

# Find the index of laptop

```python
products.index(' Laptop ')
```

# Print all the elments from second 

```python
products[1:]
```

# Print every second product 

```python
products[1::2]
```

# Add a new product "Wallet" in list

```python
products.append('wallet')
```

```python
print(products)
```

# Remove  wallet from the products

```python
products.remove('wallet')
```

```python
print(products)
```

# Extending the list with two moe products 

```python
products.extend(['Bag' , 'Bottle'])
```

```python
print(products)
```
