```python
# Create a sample dictionary

release_year_dict = {"Thriller": "1982", "Back in Black": "1980", \
                    "The Dark Side of the Moon": "1973", "The Bodyguard": "1992", \
                    "Bat Out of Hell": "1977", "Their Greatest Hits (1971-1975)": "1976", \
                    "Saturday Night Fever": "1977", "Rumours": "1977"}
release_year_dict
```

```python
# Get value by key

release_year_dict['Thriller'] 
```

```python
release_year_dict['The Bodyguard']
```

# Get all the values in the dictionary

```python
release_year_dict.values()
```

# Append the value in the dictionary through the key

```python

release_year_dict['Peaky Blinders'] = 2019
release_year_dict
```

# Delete the values in the dictionay through keys

```python
del(release_year_dict['Thriller'])
del(release_year_dict['Peaky Blinders'])
release_year_dict
```


# Show all the Keys of Dictionary 

```python
release_year_dict.keys()
```
