# Sets Logic Operations 

```python
# Creating Sample sets 

setA = {'Mobile', 'Bag', 'Laptop'}
setB = {'Mobile' , 'Laptop' , 'Headphones'}
```

```python
# print two sets

print(setA)
print(setB)
```

```python
# Find the intersection

intersection = setA & setB
intersection
```

```python
# Find the difference in setA

difference = setA.difference(setB)
difference
```

```python
# Find the difference in setB

difference = setB.difference(setA)
difference
```

```python
# Find the UNION of two sets 

setA.union(setB)
```

```python
# check superset

setA.issuperset(setB)
```

```python
# Check subset

setB.issuperset(setA)
```
