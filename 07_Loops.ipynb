```python
# For Loop example
# Changing the color of all sqaures to the white 

squares = ['red', 'yellow', 'green', 'purple', 'blue']

for square in squares:
    square = "white"
    print(square)
```

```python
# Loop through the list and iterate on both index and element value

squares=['red', 'yellow', 'green', 'purple', 'blue']

for i, square in enumerate(squares):
    print(i, square)
```

# Break Statement

```python
# Using break statement

for num in range(1, 10):
    if num == 5:
        print("Breaking the loop at:", num)
        break
    print(num)
```

# Continue Statement 

```python
# using Continue statement

for num in range(1, 6):
    if num == 3:
        continue
    print(num)
```

```python
# Using Continue and break statement together

count = 0
while count < 10:
    count += 1
    if count == 3:
        continue  # skip printing 3
    if count == 8:
        break     # stop the loop when count is 8
    print(count)
```

```python
# Continue and break statement together 
# print counting by skiping the multiples of 3 ad stop after 12 


for i in range(1, 16):
    if i % 3 == 0:
        continue  # skip multiples of 3
    if i > 12:
        break     # stop if number > 12
    print(i)
```

# While Loop

```python
# Print counting form 1 to 10

i = 1 
while i <=10:
    print(i)
    i = i+1
```

```python
# While Loop Example

dates = [1982, 1980, 1973, 2000]

i = 0
year = dates[0]

while(year != 1973):    
    print(year)
    i = i + 1
    year = dates[i]
    

print("It took ", i ,"repetitions to get out of loop.")
```

```python
# While Loop example
# break the loop where the rating is less then 6


PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]
i = 0 
rating = PlayListRatings[0]

while i<=8:
    if rating < 6:
        break
    print(rating)
    i = i+1
    rating = PlayListRatings[i]
```
